# Research Objective

## Introduction

On 16 March 2020 the daytime speed limit on most Dutch motorways was reduced to 100 km/h between 06:00 and 19:00. Before this change, the limit on many of these sections was 120 or 130 km/h, and these higher limits still apply in the evening and at night (Barendrecht & De Jong, 2021; Rijkswaterstaat, 2019).

The measure was introduced to reduce nitrogen deposition in Natura 2000 areas, not to improve road safety. Lower driving speeds are, however, one of the best-established determinants of road safety. According to the Power Model, the number of crashes changes more than proportionally with mean speed, and the more severe the crash, the stronger this effect (Elvik, 2009; Nilsson, 2004). A lower limit is therefore expected to reduce injury crashes relatively more than material-damage crashes. The policy therefore offers a large-scale, real-world setting to study how a speed-limit reduction relates to traffic safety on motorways.

Evaluating this is not straightforward, for two reasons. First, the reduction coincided with the first COVID-19 lockdown, which changed traffic volumes and travel behaviour substantially (Barendrecht & De Jong, 2021). A simple before/after comparison of accident counts would mostly capture this drop in traffic. Second, the open version of the national accident register (hereinafter referred to as: BRON) does not contain the date or time of an accident (see *Data Used*), so accidents cannot be split into those that happened under the 100 km/h regime (06:00–19:00) and those that happened under the old limit. This project addresses both issues explicitly: we correct for traffic exposure, compare affected motorway sections with sections whose limit did not change, and use the recorded light condition as a proxy for time of day.

## Objective

The objective of this project is to quantify how registered traffic safety on Dutch motorway main carriageways changed after the introduction of the 100 km/h daytime speed limit, while accounting for changes in traffic volume and for nationwide trends (such as COVID-19 and changes in accident registration) that also affected motorways where the limit did not change.

## Research questions

**Main research question**

> How did traffic safety on Dutch motorways change after the daytime speed limit was reduced from 120/130 km/h to 100 km/h in March 2020, compared with motorway sections where the speed limit did not change?

**Sub-questions**

1. **Accident frequency** 
How did the annual number of registered accidents on affected motorway sections change between the pre-policy period (2018–2019) and the post-policy period (2022–2025), relative to unaffected comparison sections?
2. **Accident rate** 
How did the accident rate (accidents per 100 million vehicle-kilometres) change when differences in traffic volume are taken into account?
3. **Accident severity and type** 
How did the number and share of accidents with injuries or fatalities (as opposed to material damage only) change on affected sections, relative to comparison sections, and which crash types (e.g. single-vehicle, rear-end) account for this change?
4. **Light condition as a proxy for time of day** 
Was the change on affected sections larger for accidents that occurred in daylight (largely during hours with the 100 km/h limit) than for accidents in darkness (largely during hours with the old 120/130 km/h limit)?

Together, these questions separate three things that are easily confused: a drop in accidents because there was less traffic, a drop that occurred on all motorways regardless of the policy, and a change that is specific to the sections and hours where the speed limit was actually lowered.

*Scope note.* Because the data do not support it (see *Data Used*), we do not claim to estimate a strict causal effect, and we do not directly compare accidents between 06:00–19:00 and 19:00–06:00. Sub-question 4 is designed as the closest feasible approximation of that comparison.

## Geographical and temporal scope

### Geographical scale

The study covers the national motorway network managed by Rijkswaterstaat (road authority type “R” in the National Road Database, NWB), restricted to motorways (route letter A) and to the **main carriageway** (*hoofdrijbaan*, `BST_CODE = HR`). National N-roads, which were not part of the policy, as well as ramps, connector roads and parallel carriageways are excluded, because their speed limits and traffic volumes are not comparable. This leaves about 4,950 km of carriageway on 41 motorways, with about 20,700 (2020) and 30,000 (2023) registered accidents.

Motorway sections are divided into two groups:

- **Treated sections:** the limit changed from 120 or 130 km/h to 100 km/h during the day (06:00–19:00), with 120/130 km/h remaining in the evening and at night.
- **Comparison sections:** the limit was already 100 km/h (at all hours) before March 2020 and did not change. Sections with lower limits (e.g. 80 km/h ring roads) are excluded from the main analysis and used only in a sensitivity check.

**Preliminary group sizes.** A first classification of the INWEVA sections, based on the speed limits recorded with the 2020 accidents, shows that both groups are large enough for the analysis. Sections with too few accidents that lie between two sections of the same group are given that group. The final classification uses the 2018–2019 data and the NDW speed-limit dataset (see *Data Used*).

| Group (preliminary, based on BRON 2020) | Length | Accidents in 2020 | Of which injury / fatal |
|---|---|---|---|
| Treated (120/130 km/h recorded) | ≈ 3,990 km | ≈ 12,600 | ≈ 430 |
| Comparison candidates (only 100 km/h recorded) | ≈ 580 km | ≈ 6,100 | ≈ 210 |
| Uncertain (too few accidents to classify) | ≈ 300 km | ≈ 1,000 | ≈ 50 |
| Limit below 100 km/h (excluded) | ≈ 90 km | ≈ 800 | ≈ 30 |

The comparison sections are mainly urban ring roads and corridors (A4, A15, A10, A16, A20, A12). They carry about 70% more traffic per kilometre than the treated sections and are more urban in character. This makes it important to compare *changes* within each group rather than levels between the groups (see *Data Pipeline*, Step 6).

### Temporal scale

The study period is **2018–2025**, determined by the overlapping availability of BRON and of INWEVA traffic volumes on a consistent road network. Because BRON only provides the year of an accident, the temporal resolution of the analysis is **one year**.

| Period | Years | Role in the analysis |
|---|---|---|
| Pre-policy | 2018–2019 | Baseline; also used to check whether treated and comparison sections followed parallel trends before the policy. |
| Transition / COVID-19 | 2020–2021 | Shown in all time series but excluded from the main before/after comparison (policy start mid-March 2020 and strongly disrupted traffic). |
| Post-policy | 2022–2025 | Main “after” period. |

## BRON: Registered traffic accidents

Data: <https://verkeersveiligheid.ndw.nu/dataset/bestand-geregistreerde-ongevallen-in-nederland-bron> and <https://downloads.rijkswaterstaatdata.nl/bron/>

The *Bestand geRegistreerde Ongevallen Nederland* (BRON) contains registered accidents linked to the NWB road network (Nationaal Dataportaal Wegverkeer, n.d.-a; Rijkswaterstaat, n.d.-a). Yearly files are available for 2003–2025. It is the main dataset of this project and is used for accident counts, severity, location and light condition.

**Data availability check.** We inspected the 2020 and 2023 open-data files. To comply with the GDPR (AVG), the open release does not contain variables that could identify individuals: all date and time fields are empty. The table below summarises the fields that matter for this project.

| Variable | Meaning | Filled 2020 | Filled 2023 |
|---|---|---|---|
| `JAAR_VKL` | Year of accident | 100% | 100% |
| `DATUM_VKL`, `MND_NUMMER`, `WEEKNR`, `DAG_CODE` | Date, month, week, weekday | 0% | 0% |
| `TIJDSTIP`, `UUR`, `DDL_ID` | Time, hour, part of day | 0% | 0% |
| `AP3_CODE` | Severity: fatal / injury / material damage only | 100% | 100% |
| `ANTL_DOD`, `ANTL_GZH`, … | Number of casualties | 0% | 0% |
| `LGD_ID` | Light condition: daylight / darkness / twilight | 100% | 100% |
| `MAXSNELHD` | Speed limit at accident location | 93% | 91% |
| `AOL_ID` | Crash type (rear-end, single-vehicle, side, …) | 100%\*\* | 100%\*\* |
| `WVK_ID` + `HECTOMETER` | Road section (NWB) and hectometre position | 71% / 30%\* | 74% / 34%\* |

\* Percentages over all accidents. For accidents on motorway main carriageways the hectometre position is filled for 94% (2020) and 96% (2023).
\*\* Always filled, but the value is “unknown” for about 70% of motorway accidents. Crash type is known for 95% of injury accidents and for only about 25% of material-damage accidents.

**Consequences for the research design.**
1. Accidents cannot be aggregated by month or by hour, so the unit of analysis becomes *road section × year*.
2. A direct day/night comparison is not possible; instead, the fully filled light-condition variable is used as a proxy (sub-question 4).
3. Severity is analysed with the three-class `AP3_CODE`; since fatal accidents on motorways are rare (about 40–45 per year), fatal and injury accidents are combined into one “injury accidents” category.
4. `MAXSNELHD` follows the daytime limit in force: in 2020, 47% of motorway accidents still carry 120 or 130 km/h, whereas in 2023 this is below 1% overall and below 3% for accidents in darkness, even though the higher night limit applies then. The field is therefore only used to classify sections with the pre-policy years 2018–2019, not to tell day from night.
5. Crash type is only analysed for injury accidents, where it is almost always known.

A more detailed BRON extract with dates is only available to researchers under a formal data-processing agreement with Rijkswaterstaat, which is not feasible within this course.

## INWEVA: Traffic intensity on road sections

Data: <https://downloads.rijkswaterstaatdata.nl/inweva/>

INWEVA (*INtensiteit WEgVAkken*) provides yearly traffic volumes for every carriageway section (*baanvak*) of the national road network (Rijkswaterstaat, n.d.-b). We downloaded and inspected the releases for 2019–2025. The 2019 release is summarised below; the differences between releases follow after the table.

| Content (INWEVA 2019) | What it contains / what we found |
|---|---|
| Sections | 8,770 sections, of which 4,394 are main carriageway (HR, including national N-roads): about 7,900 km on 60 road numbers. |
| Location fields | Road number, direction, hectometre from/to, carriageway type, section length in metres. |
| `NwbId` | The NWB road-section IDs that make up each section. This is the same key that BRON uses (`WVK_ID`), so accidents can be linked directly. |
| Daily volumes | Per day type (average day of the week, working day, weekend day), for all vehicles and per vehicle class (passenger cars, medium and heavy trucks), with lower and upper bounds. |
| Period volumes | Day (07–19 h), evening (19–23 h) and night (23–07 h). |
| Hourly volumes | Separate files with the average volume for each of the 24 hours of the day, per section and vehicle class. |
| Data quality | For HR sections: 63% measured, 12% derived, 22% traffic model, 3% manual correction (column `Bron_AL`), plus a reliability score of 1–6. |
| Year-to-year linking | A table that links each section to its section number in the previous year (folder *Netwerkmutaties*). |

**Release formats.** The releases differ in ways that determine how they are used:

| Releases | Format and network | Use in this project |
|---|---|---|
| 2019–2023 | Excel files on the VIAS network. Since 2021 each release also contains a *historie* file. | The **2023 historie file** gives volumes for **2018–2023 on one and the same network** (4,316 HR sections, all with values for every year). This is the main exposure source for 2018–2023, so no year-to-year linking of sections is needed. |
| 2024–2025 | New CSV format on the SHiVi network (*verkeersbanen*), produced with a new method by NDW and Rijkswaterstaat. | Linked to the VIAS sections through the column `vias_baan_ids` (96% of HR sections) and to BRON through `nwb_ids` (100%). |
| Every year | Hourly volume table (24 hours, per vehicle class). | Daytime/night-time and daylight/darkness exposure shares (Step 6). |

**Link test with BRON.** Accidents are linked to the sections of the 2023 network in two steps: first through `NwbId` (since 2020 each ID carries a suffix H or T that is removed), then, for accidents whose NWB ID no longer exists, on road number, direction and hectometre position. With this method **99.1% (2020) and 99.5% (2023)** of motorway accidents are linked to a section. Unlinked accidents are reported per year.

**Function: traffic exposure.** For each section and year we compute vehicle-kilometres = average daily volume (all days of the week) × 365 × section length. This is the denominator of the accident rate and the exposure term in the regression model. The hourly volumes add a time dimension that BRON lacks: **about 81% of all vehicle-kilometres on motorway main carriageways are driven between 06:00 and 19:00**, the hours of the 100 km/h limit. The policy therefore applies to the large majority of motorway exposure, even though accidents cannot be dated. The historie file also shows the size of the COVID-19 shock: the length-weighted average daily volume on motorway main carriageways fell from about 34,300 vehicles in 2019 to 27,900 in 2020 (−19%) and recovered to 33,300 in 2023. Rijkswaterstaat reports a comparable drop of about 17% (Rijkswaterstaat, 2021). Comparing raw accident counts without exposure correction would therefore be misleading.

Each BRON download contains the matching version of the NWB road network (road sections, hectometre points and road geometry as shapefile). **Function:** linking accidents to motorway number, carriageway type (HR), route letter (A), driving direction and hectometre position, and drawing maps.

## NDW speed limits and Rijkswaterstaat traffic decision

Data: <https://verkeersveiligheid.ndw.nu/dataset/snelheidslimieten> and <https://open.rijkswaterstaat.nl/open-overheid/besluiten-algemene-strekking/verkeersbesluiten/verkeersbesluit-17-december-2019/>

The NDW speed-limit dataset gives the current limit per road section, including time-dependent limits (Nationaal Dataportaal Wegverkeer, n.d.-b). Sections with 100 km/h during 06:00–19:00 and a higher limit at night are candidates for the treated group; sections with a constant 100 km/h are candidates for the comparison group. Because this dataset reflects the situation after 2020, the classification is verified with the **pre-2020 limits recorded in BRON** (`MAXSNELHD`, 2018–2019): a comparison section must already show 100 km/h before the policy, and a treated section 120 or 130 km/h. Appendix A of the traffic decision of 17 December 2019 (Rijkswaterstaat, 2019) is used to spot-check the result.

# Data Pipeline

All steps are implemented in Python (pandas, statsmodels, matplotlib) and documented in this notebook. Reusable functions are in the module `pipeline/` and two scripts make the pipeline reproducible: `scripts/download_data.py` downloads the raw data and `scripts/build_dataset.py` builds the analysis tables (see the README).

### Step 1: Data collection

- `download_data.py` downloads BRON (2018–2025) and INWEVA (2023–2025) from the Rijkswaterstaat download server; the INWEVA 2023 historie file already covers 2018–2023. The NDW speed-limit file is downloaded separately.
- Raw data is not committed to GitHub (several GB); the repository contains the scripts and the processed, aggregated dataset.

### Step 2: Cleaning and filtering

- Read `ongevallen.txt` (latin-1) per year and join it with that year’s NWB road sections (`wegvakken.txt`) via `WVK_ID`.
- Keep Rijkswaterstaat motorway main carriageways (`WEGBEHSRT = R`, `BST_CODE = HR`, `ROUTELTR = A`).
- Recode severity (`AP3_CODE`: DOD/LET → injury accident, UMS → material damage only), light condition (`LGD_ID`: daylight, darkness, twilight) and crash type (`AOL_ID`).
- Accidents that BRON links to a junction instead of a road section are excluded: this concerns about 1% of the accidents on main carriageways.

### Step 3: Defining treated and comparison sections

- Use the INWEVA section of the 2023 network as the spatial unit. For 2018–2023 the volumes come directly from the 2023 historie file. The 2024–2025 SHiVi sections are mapped onto these sections via `vias_baan_ids` (length-weighted where several SHiVi sections cover one VIAS section). This gives a volume for 98% (2024) and 97.5% (2025) of the motorway sections; the rest are left out for those years.
- Assign each section to treated / comparison / excluded using the pre-2020 `MAXSNELHD` values and the NDW speed-limit data (see *Data Used*).
- INWEVA sections are short (median about 4 accidents per section per year). For the model they are combined into longer homogeneous stretches with the same group, road number and direction (about 425 stretches, on average 12 km long), to avoid many zero counts.

### Step 4: Data integration

Each accident is linked to a section of the 2023 network (NWB ID, then hectometre position; see *Data Used*). Accident counts are combined with the INWEVA volume of the same section and year, and aggregated to the analytical dataset. The rows below are illustrative:

| A2 Re 120–135 | 2019 | Before | Treated | … | … | … | … / … |
| A2 Re 120–135 | 2023 | After | Treated | … | … | … | … / … |
| A10 Li 20–32 | 2019 | Before | Comparison | … | … | … | … / … |

### Step 5: Exploratory data analysis

- Accident counts, accident rates, severity and light condition over time, for treated vs comparison sections.
- Pre-trend check: did both groups develop similarly in 2018–2019? This is the key assumption of the comparison in Step 6.
- Missing values, outliers and sections with implausible volumes.

### Step 6: Statistical analysis

**Difference-in-differences.** The change on treated sections (after minus before) is compared with the change on comparison sections. Nationwide influences such as COVID-19-related traffic changes, weather and changes in police registration affect both groups and therefore cancel out.

**Count regression with exposure (sub-questions 1–2).** Accident counts per section (or stretch) and year are modelled with a Poisson or negative binomial regression (statsmodels GLM):

$$\log E[\text{accidents}_{it}] = \log(\text{vehicle-km}_{it}) + \alpha_i + \gamma_t + \beta \cdot (\text{treated}_i \times \text{post}_t)$$

Here $\alpha_i$ are section effects, $\gamma_t$ year effects, and $\exp(\beta)$ is the incidence-rate ratio: the relative change in the accident rate on treated sections compared with comparison sections. An event-study version (separate interactions per year) shows the full development including 2020–2021.

**Severity and type (sub-question 3).** The same model is estimated for injury accidents only, and the share of injury accidents is compared between groups and periods. Following the Power Model, we expect a larger relative reduction for injury accidents than for material-damage accidents. For injury accidents, the change is further broken down by crash type. This part is descriptive, because the numbers per type are small (a few hundred injury accidents per year in the treated group).

**Light-condition proxy (sub-question 4).** The model is estimated separately for daylight and darkness accidents (a triple-difference). In summer, darkness falls almost entirely outside 06:00–19:00, but on winter mornings and late afternoons it overlaps with the 100 km/h hours; conversely, summer evenings after 19:00 are light. Both kinds of misclassification push the daylight and darkness estimates towards each other, so a difference that is still found can be interpreted as a conservative (lower-bound) indication.

To quantify how good this proxy is, the INWEVA hourly volumes are combined with sunrise and sunset times for every day of the year (computed in Python). This gives, per section, an estimate of the share of vehicle-km driven in daylight and in darkness, and of how much of each falls within 06:00–19:00. These shares serve as exposure for the daylight and darkness models and show how strongly each light category is affected by the policy. The assumption is that the hourly traffic profile is the same throughout the year.

# Expected Limitations

- **No date or time in open BRON.** The analysis is limited to yearly totals, and the day/night distinction relies on light condition as an imperfect proxy (*Data Pipeline*, Step 6).
- **COVID-19.** The policy started at the same moment as the first lockdown. Using comparison sections, exposure correction and a separate transition period limits, but does not fully remove, this problem: COVID-19 may have affected busy urban 100 km/h sections differently from rural 130 km/h sections.
- **Comparability of groups.** Comparison sections are often near cities, with more congestion and junctions. The section effects in the model absorb permanent differences; the pre-trend check tests whether trends were similar.
- **Registration.** BRON only contains registered accidents, and registration practices change over time (total registrations rose from about 104,000 in 2020 to 134,000 in 2023, and injury accidents on motorway main carriageways from about 730 to 1,420). Changes that affect all motorways equally are absorbed by the comparison group.
- **Exposure data quality.** About a third of INWEVA motorway volumes are derived or modelled rather than measured. A sensitivity analysis uses only sections with measured volumes.
- **Method changes in INWEVA.** According to the release notes, the estimation method changed in 2020 (new contractor) and again in 2024 (switch to SHiVi data on a new network). Both can cause local trend breaks. Rijkswaterstaat reports that for about 90% of the network length the new method differs by less than ±5% from the old one (Rijkswaterstaat, 2025). Because the breaks affect treated and comparison sections alike, the comparison-group design largely absorbs them. As a robustness check, the analysis is repeated with 2018–2023 only (one consistent historie file) and with sections that have measured volumes.
- **Route choice.** Rijkswaterstaat notes that the lower limit may have shifted some traffic to non-motorway roads (Rijkswaterstaat, 2021). Such shifts change motorway exposure, which INWEVA captures, but the effect on safety elsewhere in the network is outside the scope of this project.
- **Small numbers for severe accidents.** Fatal accidents are too rare for a separate section-level analysis and are combined with injury accidents.
- **Speed limit ≠ actual speed.** Without the optional NDW extension we cannot show how much actual speeds changed; we rely on the Rijkswaterstaat monitoring for that context.
- **Association, not proof of causation.** The design strengthens the interpretation but remains observational.
